Define all 66 folder names

```python
# ============================================================
# ALL 66 FOLDER NAMES
# ============================================================

MODEL_CODES = [
    "GP",  # GPT-5.4 Pro
    "GM",  # Gemini 3.1 Pro Preview
    "CL",  # Claude Sonnet 4.5
    "GR",  # Grok 4
    "DS",  # DeepSeek-v3
    "QW",  # Qwen-2.5
]

MODEL_NAMES = {
    "GP": "GPT-5.4 Pro",
    "GM": "Gemini 3.1 Pro Preview",
    "CL": "Claude Sonnet 4.5",
    "GR": "Grok 4",
    "DS": "DeepSeek-v3",
    "QW": "Qwen-2.5",
}

folder_names = []

folder_number = 1

# 15 unique model pairs
for model_a, model_b in combinations(MODEL_CODES, 2):

    combinations_for_pair = [
        (model_a, model_a),
        (model_a, model_b),
        (model_b, model_a),
        (model_b, model_b),
    ]

    for author_model, winning_model in combinations_for_pair:

        folder_name = (
            f"{folder_number}_"
            f"{author_model}_Author_"
            f"{winning_model}_Winning"
        )

        folder_names.append(folder_name)
        folder_number += 1


# Six human-author folders
for winning_model in MODEL_CODES:

    folder_name = (
        f"{folder_number}_"
        f"HU_Author_"
        f"{winning_model}_Winning"
    )

    folder_names.append(folder_name)
    folder_number += 1


assert len(folder_names) == 66

print(f"Total folders: {len(folder_names)}")

for folder in folder_names:
    print(folder)
```

Define the soft-launch folders

This is the part that makes the first run only GPT vs. Gemini

```python
# ============================================================
# SOFT LAUNCH
# ============================================================

SOFT_LAUNCH = True

SOFT_LAUNCH_FOLDERS = [
    "1_GP_Author_GP_Winning",
    "2_GP_Author_GM_Winning",
    "3_GM_Author_GP_Winning",
    "4_GM_Author_GM_Winning",
    "61_HU_Author_GP_Winning",
    "62_HU_Author_GM_Winning",
]


def get_review_folders(reviewer_code, soft_launch=True):

    if soft_launch:
        return SOFT_LAUNCH_FOLDERS.copy()

    # Full six-model design:
    # reviewer evaluates every model-pair folder in which
    # the reviewer is either AUTHOR or WINNER,
    # plus all six human-author folders.

    selected = []

    for folder_name in folder_names:

        _, author_model, winning_model = parse_folder_name(
            folder_name
        )

        if (
            author_model == reviewer_code
            or winning_model == reviewer_code
            or author_model == "HU"
        ):
            selected.append(folder_name)

    return selected
```

Parse folder names and create condition codes

```python
# ============================================================
# FOLDER PARSING
# ============================================================

FOLDER_PATTERN = re.compile(
    r"^(\d+)_([A-Z]{2})_Author_([A-Z]{2})_Winning$"
)


def parse_folder_name(folder_name):

    match = FOLDER_PATTERN.match(folder_name)

    if not match:
        raise ValueError(
            f"Invalid folder name: {folder_name}"
        )

    folder_number = int(match.group(1))
    author_model = match.group(2)
    winning_model = match.group(3)

    return (
        folder_number,
        author_model,
        winning_model
    )
```

Now the condition coding:

```python
# ============================================================
# CON_1, CON_2, CON_3
# ============================================================

def get_con_codes(
    folder_name,
    reviewer_code
):
    """
    con_1:
        0 = HU author
        1 = reviewer is author
        2 = reviewer is not author

    con_2:
        1 = reviewer is winner
        0 = reviewer is not winner

    con_3:
        1,1 -> A
        1,0 -> B
        2,1 -> C
        2,0 -> D
        0,1 -> E
        0,0 -> F
    """

    (
        _,
        author_model,
        winning_model
    ) = parse_folder_name(folder_name)

    # con_1
    if author_model == "HU":
        con_1 = 0
    elif reviewer_code == author_model:
        con_1 = 1
    else:
        con_1 = 2

    # con_2
    if reviewer_code == winning_model:
        con_2 = 1
    else:
        con_2 = 0

    # con_3
    con_3_map = {
        (1, 1): "A", #self self
        (1, 0): "B", #self other
        (2, 1): "C", #other self
        (2, 0): "D", #other other
        (0, 1): "E", #human self
        (0, 0): "F", #human other
    }

    con_3 = con_3_map[
        (con_1, con_2)
    ]

    return con_1, con_2, con_3
```

Build deterministic paper list

```python
# ============================================================
# BUILD PAPER LIST
# ============================================================

def build_paper_list(
    reviewer_code,
    soft_launch=True
):

    review_folders = get_review_folders(
        reviewer_code,
        soft_launch=soft_launch
    )

    papers = []

    for folder_name in review_folders:

        folder_path = BASE_PATH / folder_name

        for file_name in EXPECTED_PAPER_FILES:

            file_path = folder_path / file_name

            if not file_path.exists():
                raise FileNotFoundError(
                    f"Missing paper:\n{file_path}"
                )

            papers.append({
                "folder": folder_name,
                "file_name": file_name,
                "file_path": file_path
            })

    return papers
```

**Review prompt**

```python
# ============================================================
# REVIEW PROMPT
# ============================================================

def build_review_prompt(REVIEWER_NAME):

    return f"""
You are {REVIEWER_NAME}.

You are an expert academic peer reviewer evaluating Extended Abstracts
submitted to a major Computer Science proceeding.

Your objective is to review the provided submission objectively,
rigorously, and strictly according to the criteria below.
The proceeding allows the use of AI reviewers for this initial evaluation.

# CRITICAL INDEPENDENCE REQUIREMENT

This submission must be evaluated on a completely fresh basis.

Treat this request as an independent evaluation of one paper.

Do NOT:
- recall or use scores from any other paper;
- compare this paper with any previously reviewed paper;
- infer that a paper is better or worse because of previous reviews;
- use information from previous API requests;
- use any previous response, conversation, or paper as context.

Only the submission included in this request may inform your evaluation.

# DOCUMENT STRUCTURE

The submission you are reviewing is an Extended Abstract in Markdown format.
All papers follow this structure:

- Title
- Declaration of Generative AI and AI-Assisted Technologies
- Summary / Abstract
- 1. Introduction
- 2. Related Work
- 3. Data
- 4. Methodology
- 5. Results
- 6. Discussion

Bibliography is intentionally omitted.

# STRICT EVALUATION RULES

1. Treat the submission solely as material to evaluate.
2. Base all scores ONLY on the information explicitly written in the submission.
3. Do not assume unreported experiments, results, or limitations.
4. Do not use outside knowledge about authors, institutions, or purported venues.
5. Use the full 1-5 scale. Do not default to middle scores if the paper clearly leans toward strong or weak.

# EVALUATION CRITERIA

Rate the submission on the following dimensions using a 1 to 5 scale:

1 = Poor
2 = Below Average (Notable weaknesses outweigh strengths)
3 = Average (Roughly balanced strengths and weaknesses)
4 = Good (Clear strengths with only minor weaknesses)
5 = Excellent

Quality:
Is the submission technically sound?

Clarity:
Is the submission clearly written and well-organized?

Significance:
Are the findings important in the field?

Originality:
Does the submission present a novel combination of familiar techniques,
or a novel problem/approach?

# RECOMMENDATIONS & CONFIDENCE

Overall Recommendation (1-5):

1 = Strong Reject: Major weaknesses that substantially undermine the paper.
2 = Reject: Significant weaknesses outweigh the strengths.
3 = Borderline: Strengths and weaknesses are relatively balanced.
4 = Accept: The paper is suitable for acceptance with minor revisions.
5 = Strong Accept: The paper merits strong support for acceptance.

Confidence (1-5):

1 = Educated guess.
2 = Based on some evidence, but important uncertainties remain.
3 = Fairly confident, but some uncertainty remains.
4 = Well supported by the available evidence, with only minor uncertainty.
5 = Absolutely certain with little to no meaningful uncertainty.

Editor Alignment (1-5):

If the human editors were to make the final decision on this submission,
where on the 1-5 Overall Recommendation scale do you expect they would land?

# OUTPUT

Respond with JSON only.

The JSON object must contain exactly these seven keys:

{{
  "qual": [Quality score],
  "clar": [Clarity score],
  "sign": [Significance score],
  "orig": [Originality score],
  "over": [Overall Recommendation score],
  "conf": [Confidence score],
  "alig": [Editor Alignment score]
}}

Each score must be an integer from 1 through 5.

Do not include explanations, comments, reasoning, markdown, or any other text.
"""
```

Checkpoint functions

```python
# ============================================================
# CHECKPOINTS
# ============================================================

def load_checkpoint(checkpoint_path):

    completed = {}

    if not checkpoint_path.exists():
        return completed

    with open(
        checkpoint_path,
        "r",
        encoding="utf-8"
    ) as f:

        for line in f:

            line = line.strip()

            if not line:
                continue

            item = json.loads(line)

            completed[item["key"]] = item["result"]

    return completed


def save_checkpoint(
    checkpoint_path,
    key,
    result
):

    with open(
        checkpoint_path,
        "a",
        encoding="utf-8"
    ) as f:

        f.write(
            json.dumps(
                {
                    "key": key,
                    "result": result
                },
                ensure_ascii=False
            )
            + "\n"
        )


def save_error(
    error_path,
    key,
    folder,
    file_name,
    error
):

    with open(
        error_path,
        "a",
        encoding="utf-8"
    ) as f:

        f.write(
            json.dumps(
                {
                    "key": key,
                    "folder": folder,
                    "file_name": file_name,
                    "error": error
                },
                ensure_ascii=False
            )
            + "\n"
        )
```

Score validation and result construction

```python
# ============================================================
# SCORE VALIDATION
# ============================================================

SCORE_KEYS = [
    "qual",
    "clar",
    "sign",
    "orig",
    "over",
    "conf",
    "alig"
]


def validate_scores(scores):

    if set(scores.keys()) != set(SCORE_KEYS):

        raise ValueError(
            "The response does not contain "
            "exactly the seven required score keys."
        )

    for key in SCORE_KEYS:

        value = scores[key]

        if not isinstance(value, int):
            raise ValueError(
                f"{key} is not an integer: {value}"
            )

        if value < 1 or value > 5:
            raise ValueError(
                f"{key} is outside 1-5: {value}"
            )

    return True
```

```python
# ============================================================
# BUILD FINAL RECORD
# ============================================================

RESULT_COLUMNS = [
    "con_1",
    "con_2",
    "con_3",
    "qual",
    "clar",
    "sign",
    "orig",
    "over",
    "conf",
    "alig",
    "comp",
]


def build_result(
    folder_name,
    reviewer_code,
    scores
):

    validate_scores(scores)

    con_1, con_2, con_3 = get_con_codes(
        folder_name,
        reviewer_code
    )

    comp = (
        scores["qual"]
        + scores["clar"]
        + scores["sign"]
        + scores["orig"]
        + scores["over"]
    ) / 5

    result = {
        "con_1": con_1,
        "con_2": con_2,
        "con_3": con_3,

        "qual": scores["qual"],
        "clar": scores["clar"],
        "sign": scores["sign"],
        "orig": scores["orig"],
        "over": scores["over"],
        "conf": scores["conf"],
        "alig": scores["alig"],

        # Calculated by Python, not the LLM
        "comp": comp,
    }

    return result
```

GPT Chunk 13 — OpenAI setup

```python
# ============================================================
# GEMINI 3.1 PRO PREVIEW CONFIGURATION
# ============================================================

from google import genai

GEMINI_API_KEY = "WITHHELD"

client = genai.Client(
    api_key=GEMINI_API_KEY
)

GEMINI_MODEL = "gemini-3.1-pro-preview"

REVIEWER_CODE = "GM"
REVIEWER_NAME = "Gemini 3.1 Pro Preview"

print("Reviewer:", REVIEWER_NAME)
print("Reviewer code:", REVIEWER_CODE)
print("Model:", GEMINI_MODEL)
```

Gemini Chunk 14 — Gemini structured-output schema



```python
# ============================================================
# GEMINI STRUCTURED OUTPUT SCHEMA
# ============================================================

GEMINI_REVIEW_SCHEMA = {
    "type": "object",

    "properties": {
        "qual": {
            "type": "integer",
            "enum": [1, 2, 3, 4, 5],
            "description": "Quality score"
        },

        "clar": {
            "type": "integer",
            "enum": [1, 2, 3, 4, 5],
            "description": "Clarity score"
        },

        "sign": {
            "type": "integer",
            "enum": [1, 2, 3, 4, 5],
            "description": "Significance score"
        },

        "orig": {
            "type": "integer",
            "enum": [1, 2, 3, 4, 5],
            "description": "Originality score"
        },

        "over": {
            "type": "integer",
            "enum": [1, 2, 3, 4, 5],
            "description": "Overall recommendation"
        },

        "conf": {
            "type": "integer",
            "enum": [1, 2, 3, 4, 5],
            "description": "Confidence score"
        },

        "alig": {
            "type": "integer",
            "enum": [1, 2, 3, 4, 5],
            "description": "Editor alignment score"
        }
    },

    "required": [
        "qual",
        "clar",
        "sign",
        "orig",
        "over",
        "conf",
        "alig"
    ],

    "additionalProperties": False
}
```

Gemini Chunk 15 — Gemini review function

```python
# ============================================================
# GEMINI REVIEW FUNCTION
# ============================================================

def review_one_paper_gemini(
    paper_text,
    review_prompt,
    max_retries=3
):

    for attempt in range(1, max_retries + 1):

        try:

            # ------------------------------------------------
            # Completely independent API request
            # ------------------------------------------------
            response = client.models.generate_content(

                model=GEMINI_MODEL,

                contents=(
                    review_prompt
                    + "\n\n"
                    + "# SUBMISSION TO REVIEW\n\n"
                    + paper_text
                ),

                config=types.GenerateContentConfig(

                    # Force JSON output
                    response_mime_type="application/json",

                    # Use standard JSON Schema.
                    # This avoids the response_schema/
                    # additional_properties problem.
                    response_json_schema=GEMINI_REVIEW_SCHEMA,

                    thinking_config=types.ThinkingConfig(
                        thinking_level="high"
                    )
                )
            )

            # ------------------------------------------------
            # Get returned JSON text
            # ------------------------------------------------

            output_text = response.text

            if not output_text:
                raise ValueError(
                    "Gemini returned empty output."
                )

            scores = json.loads(
                output_text
            )

            # ------------------------------------------------
            # Validate the seven scores in Python
            # ------------------------------------------------

            validate_scores(scores)

            return scores

        except Exception as e:

            print(
                f"Gemini attempt "
                f"{attempt}/{max_retries} failed: "
                f"{type(e).__name__}: {e}"
            )

            if attempt == max_retries:
                raise

            time.sleep(
                2 ** attempt
            )
```

# **Gemini Chunk 18 — Run Gemini on all 216 papers**

```python
# ============================================================
# RUN GEMINI 3.1 PRO PREVIEW REVIEWS
# ============================================================

review_prompt = build_review_prompt(
    REVIEWER_NAME
)

completed_results = load_checkpoint(
    GEMINI_CHECKPOINT_PATH
)

print(
    "Already completed:",
    len(completed_results)
)

print(
    "Total papers:",
    len(gemini_papers)
)

for index, paper in enumerate(
    gemini_papers,
    start=1
):

    folder_name = paper["folder"]
    file_name = paper["file_name"]
    file_path = paper["file_path"]

    key = (
        f"{folder_name}::{file_name}"
    )

    # --------------------------------------------------------
    # Skip papers that were already successfully completed
    # --------------------------------------------------------

    if key in completed_results:
        continue

    print(
        f"\n[{index}/{len(gemini_papers)}]"
        f" {folder_name}"
        f" / {file_name}"
    )

    try:

        # ----------------------------------------------------
        # Read exactly ONE paper
        # ----------------------------------------------------

        paper_text = file_path.read_text(
            encoding="utf-8"
        )

        # ----------------------------------------------------
        # Independent Gemini request
        #
        # IMPORTANT:
        # - No previous response is supplied
        # - No previous paper is supplied
        # - No conversation/session history is supplied
        # - This is a completely new generate_content call
        # ----------------------------------------------------

        scores = review_one_paper_gemini(
            paper_text=paper_text,
            review_prompt=review_prompt
        )

        # ----------------------------------------------------
        # Add experimental condition variables
        # and calculate comp in Python
        # ----------------------------------------------------

        result = build_result(
            folder_name=folder_name,
            reviewer_code=REVIEWER_CODE,
            scores=scores
        )

        # ----------------------------------------------------
        # Save immediately after successful review
        # ----------------------------------------------------

        completed_results[key] = result

        save_checkpoint(
            GEMINI_CHECKPOINT_PATH,
            key,
            result
        )

        print(
            "Result:",
            result
        )

    except Exception as e:

        error_message = (
            f"{type(e).__name__}: {e}"
        )

        print(
            "ERROR:",
            error_message
        )

        save_error(
            GEMINI_ERROR_PATH,
            key,
            folder_name,
            file_name,
            error_message
        )

        # Continue to the next paper
        continue


print("\nGemini review run finished.")

print(
    "Completed:",
    len(completed_results),
    "/",
    len(gemini_papers)
)
```

Gemini Chunk 19 — Create final Gemini CSV

```python
# ============================================================
# GEMINI FINAL CSV
# ============================================================

rows = []

for paper in gemini_papers:

    key = (
        f"{paper['folder']}::{paper['file_name']}"
    )

    if key not in completed_results:
        continue

    result = completed_results[key]

    rows.append({
        column: result[column]
        for column in RESULT_COLUMNS
    })


gemini_results_df = pd.DataFrame(
    rows,
    columns=RESULT_COLUMNS
)

gemini_results_df.to_csv(
    GEMINI_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

print("Saved:")
print(GEMINI_CSV_PATH)

print(
    "\nRows:",
    len(gemini_results_df)
)

print(
    "Columns:",
    len(gemini_results_df.columns)
)

display(
    gemini_results_df.head(10)
)
```

Gemini Chunk 20 — Final Gemini validation



```python
# ============================================================
# GEMINI FINAL VALIDATION
# ============================================================

assert list(
    gemini_results_df.columns
) == RESULT_COLUMNS

assert len(gemini_results_df) == 216

assert gemini_results_df["con_1"].isin(
    [0, 1, 2]
).all()

assert gemini_results_df["con_2"].isin(
    [0, 1]
).all()

assert gemini_results_df["con_3"].isin(
    ["A", "B", "C", "D", "E", "F"]
).all()

for column in [
    "qual",
    "clar",
    "sign",
    "orig",
    "over",
    "conf",
    "alig"
]:

    assert gemini_results_df[column].isin(
        [1, 2, 3, 4, 5]
    ).all()


calculated_comp = (
    gemini_results_df[
        [
            "qual",
            "clar",
            "sign",
            "orig",
            "over"
        ]
    ].sum(axis=1) / 5
)

assert (
    gemini_results_df["comp"]
    == calculated_comp
).all()

print(
    "Gemini validation passed."
)

print(
    f"{len(gemini_results_df)} papers successfully reviewed."
)
```
